```python
import pandas as pd

customers = pd.read_csv("../Downloads/archive/olist_customers_dataset.csv")
orders = pd.read_csv("../Downloads/archive/olist_orders_dataset.csv")

print("Customers Dataset:")
display(customers.head())

print("Orders Dataset:")
display(orders.head())
```

```python
print("Customers Dataset Shape:", customers.shape)
print("Orders Dataset Shape:", orders.shape)

print("\nCustomers Information:")
customers.info()

print("\nOrders Information:")
orders.info()
```

```python
print("Missing Values in Customers:")
print(customers.isnull().sum())

print("\nMissing Values in Orders:")
print(orders.isnull().sum())
```

```python
missing_report = pd.DataFrame({
    'Missing Count': orders.isnull().sum(),
    'Missing Percentage': orders.isnull().mean() * 100
})

display(missing_report)
```

```python
print("Order Status Count:")
print(orders['order_status'].value_counts())

print("\nMissing Dates by Order Status:")

status_missing = orders.groupby('order_status')[
    ['order_approved_at',
     'order_delivered_carrier_date',
     'order_delivered_customer_date']
].apply(lambda x: x.isnull().sum())

display(status_missing)
```

```python
items = pd.read_csv("../Downloads/archive/olist_order_items_dataset.csv")
products = pd.read_csv("../Downloads/archive/olist_products_dataset.csv")
payments = pd.read_csv("../Downloads/archive/olist_order_payments_dataset.csv")
reviews = pd.read_csv("../Downloads/archive/olist_order_reviews_dataset.csv")
sellers = pd.read_csv("../Downloads/archive/olist_sellers_dataset.csv")

print("All datasets loaded successfully!")

print("Items:", items.shape)
print("Products:", products.shape)
print("Payments:", payments.shape)
print("Reviews:", reviews.shape)
print("Sellers:", sellers.shape)
```

```python
datasets = {
    "Items": items,
    "Products": products,
    "Payments": payments,
    "Reviews": reviews,
    "Sellers": sellers
}

for name, df in datasets.items():
    print(name, "Duplicate Rows:", df.duplicated().sum())
```

```python
all_datasets = {
    "Customers": customers,
    "Orders": orders,
    "Items": items,
    "Products": products,
    "Payments": payments,
    "Reviews": reviews,
    "Sellers": sellers
}

for name, df in all_datasets.items():
    print("\n", name)
    print(df.isnull().sum())
```

```python
missing_summary = []

for name, df in all_datasets.items():
    for col in df.columns:
        missing_summary.append({
            "Dataset": name,
            "Column": col,
            "Missing_Count": df[col].isnull().sum(),
            "Missing_Percentage": round(df[col].isnull().mean() * 100, 2)
        })

missing_report_all = pd.DataFrame(missing_summary)

display(missing_report_all)

missing_report_all.to_csv("missing_values_report.csv", index=False)
```

```python
unique_id_columns = {
    "Customers": (customers, "customer_id"),
    "Orders": (orders, "order_id"),
    "Products": (products, "product_id"),
    "Sellers": (sellers, "seller_id"),
    "Items": (items, ["order_id", "order_item_id"]),
    "Payments": (payments, ["order_id", "payment_sequential"]),
    "Reviews": (reviews, "review_id")
}

for name, (df, column) in unique_id_columns.items():
    print(name, "Duplicate IDs:", df.duplicated(subset=column).sum())
```

```python
print("Negative Quantity:", (items["order_item_id"] < 0).sum())

print("Zero or Negative Price:", (items["price"] <= 0).sum())

print("Negative Freight Value:", (items["freight_value"] < 0).sum())

print("Negative Payment Value:", (payments["payment_value"] < 0).sum())

print("Invalid Review Scores:",
      (~reviews["review_score"].between(1, 5)).sum())
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")
    print(col, "Missing/Invalid Dates:", orders[col].isnull().sum())
```

```python
sales_data = items.merge(
    products,
    on="product_id",
    how="left"
)

sales_data = sales_data.merge(
    orders[["order_id", "customer_id", "order_status",
            "order_purchase_timestamp"]],
    on="order_id",
    how="left"
)

sales_data = sales_data.merge(
    customers[["customer_id", "customer_city", "customer_state"]],
    on="customer_id",
    how="left"
)

print("Merged Sales Dataset:", sales_data.shape)
display(sales_data.head())
```

```python
total_revenue = items["price"].sum()

total_orders = orders["order_id"].nunique()

average_order_value = total_revenue / total_orders

total_units_sold = len(items)

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Orders",
        "Average Order Value",
        "Total Units Sold"
    ],
    "Value": [
        total_revenue,
        total_orders,
        average_order_value,
        total_units_sold
    ]
})

display(kpi_summary)
```

```python
top_products = sales_data.groupby("product_id").agg(
    Total_Orders=("order_id", "count"),
    Total_Revenue=("price", "sum")
).sort_values("Total_Revenue", ascending=False)

display(top_products.head(10))
```

```python
sales_data.columns = sales_data.columns.str.strip()

category_sales = sales_data.groupby(
    "product_category_name_english"
).agg(
    Total_Revenue=("price", "sum"),
    Total_Items=("order_id", "count")
).sort_values("Total_Revenue", ascending=False)

display(category_sales.head(10))
```

```python
sales_data["order_purchase_timestamp"] = pd.to_datetime(
    sales_data["order_purchase_timestamp"]
)

sales_data["Month"] = sales_data[
    "order_purchase_timestamp"
].dt.to_period("M").astype(str)

monthly_sales = sales_data.groupby("Month")["price"].sum()

display(monthly_sales)
```

```python
customer_orders = orders.groupby("customer_id")[
    "order_id"
].nunique()

repeat_customers = (customer_orders > 1).sum()

one_time_customers = (customer_orders == 1).sum()

print("Repeat Customers:", repeat_customers)
print("One-Time Customers:", one_time_customers)
```

```python
plt.figure(figsize=(10, 5))

top_10_products = top_products.head(10)

plt.bar(
    top_10_products.index.astype(str),
    top_10_products["Total_Revenue"]
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Product ID")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("top_products_chart.png")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()

plt.savefig("monthly_sales_chart.png")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

kpi_names = [
    "Revenue",
    "Orders",
    "Average Order Value",
    "Order Items"
]

kpi_values = [
    total_revenue,
    total_orders,
    average_order_value,
    total_units_sold
]

ax.bar(kpi_names, kpi_values)

ax.set_title("E-Commerce Business KPI Dashboard")
ax.set_ylabel("Value")

plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig("ecommerce_dashboard.png")
plt.show()
```

```python
category_data = items.merge(
    products[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

# Calculate revenue by category
category_sales = category_data.groupby(
    'product_category_name'
)['price'].sum().sort_values(ascending=False)

# Select top 5 categories
top_5_categories = category_sales.head(5)

# Create Pie Chart
plt.figure(figsize=(8, 8))

plt.pie(
    top_5_categories.values,
    labels=top_5_categories.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Top 5 Product Categories by Revenue")
plt.show()
```
